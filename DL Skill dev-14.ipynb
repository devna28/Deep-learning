```python
#task14
import torch
import torch.nn as nn

# Sample Input
X = torch.randn(5)

# Custom Swish Activation
class Swish(nn.Module):

    def forward(self, x):
        return x * torch.sigmoid(x)

# Create Swish Layer
swish = Swish()

# Apply Swish
swish_output = swish(X)

# Compare with ReLU
relu = nn.ReLU()
relu_output = relu(X)

print("Input:")
print(X)

print("\nSwish Output:")
print(swish_output)

print("\nReLU Output:")
print(relu_output)
```
