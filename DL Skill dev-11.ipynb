```python
#task11
import torch
import torch.nn as nn

# Sample Image (Batch=1, Channels=3, Height=32, Width=32)
X = torch.randn(1, 3, 32, 32)

# Convert Image into Patches
patch_size = 8
patches = X.unfold(2, patch_size, patch_size).unfold(3, patch_size, patch_size)
patches = patches.contiguous().view(1, -1, 3 * patch_size * patch_size)

# Patch Projection
projection = nn.Linear(3 * patch_size * patch_size, 64)
patch_embeddings = projection(patches)

# Class Token
cls_token = torch.randn(1, 1, 64)

# Add Class Token
tokens = torch.cat((cls_token, patch_embeddings), dim=1)

# Positional Embedding
position = torch.randn(1, tokens.shape[1], 64)
tokens = tokens + position

# Multi-Head Self-Attention
attention = nn.MultiheadAttention(embed_dim=64, num_heads=4, batch_first=True)

output, weights = attention(tokens, tokens, tokens)

print("Patch Shape :", patches.shape)
print("Embedding Shape :", patch_embeddings.shape)
print("Transformer Output Shape :", output.shape)
print("Attention Shape :", weights.shape)
```
