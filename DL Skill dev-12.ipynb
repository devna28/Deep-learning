```python
#task12
import torch
import torch.nn as nn

# Sample Images
X = torch.randn(2, 1, 28, 28)

# Convolutional Autoencoder

class AutoEncoder(nn.Module):

    def __init__(self):
        super().__init__()

        # Encoder
        self.encoder = nn.Sequential(
            nn.Conv2d(1, 8, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2)
        )

        # Decoder
        self.decoder = nn.Sequential(
            nn.ConvTranspose2d(8, 1, 2, stride=2),
            nn.Sigmoid()
        )

    def forward(self, x):
        x = self.encoder(x)
        x = self.decoder(x)
        return x

model = AutoEncoder()

# Reconstructed Image
output = model(X)


# Reconstruction Loss

loss = nn.MSELoss()
reconstruction_loss = loss(output, X)

# Anomaly Score

anomaly_score = torch.mean((output - X) ** 2)

print("Input Shape :", X.shape)
print("Output Shape:", output.shape)
print("Reconstruction Loss :", reconstruction_loss.item())
print("Anomaly Score :", anomaly_score.item())
```
