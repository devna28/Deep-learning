```python
#task 1
import numpy as np

# Input and Target
X = np.array([[1,2],
              [3,4],
              [5,6]])

Y = np.array([[1,0],
              [0,1],
              [1,0]])

# Initialize Weights
W1 = np.random.rand(2,3)
W2 = np.random.rand(3,2)

lr = 0.01

# ReLU
def relu(x):
    return np.maximum(0,x)

# ReLU Derivative
def relu_derivative(x):
    return (x > 0).astype(int)

# Softmax
def softmax(x):
    exp = np.exp(x)
    return exp / np.sum(exp, axis=1, keepdims=True)

# Training
for epoch in range(100):

    # Forward Pass
    Z1 = np.dot(X, W1)
    A1 = relu(Z1)

    Z2 = np.dot(A1, W2)
    Output = softmax(Z2)

    # Loss
    loss = -np.mean(Y * np.log(Output + 1e-8))

    # Backpropagation
    dOutput = Output - Y

    dW2 = np.dot(A1.T, dOutput)

    dHidden = np.dot(dOutput, W2.T) * relu_derivative(Z1)

    dW1 = np.dot(X.T, dHidden)

    # Update Weights
    W2 = W2 - lr * dW2
    W1 = W1 - lr * dW1

print("Final Loss:", loss)

print("\nPredicted Output:")
print(Output)
```
