```python
#task2
import torch
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create Input
torch.manual_seed(42)
X = torch.randn(100, 50)

layers = 20
size = 50
```

```python
# Xavier Initialization
def xavier(in_features, out_features):
    limit = np.sqrt(6 / (in_features + out_features))
    return torch.empty(in_features, out_features).uniform_(-limit, limit)
```

```python

# Kaiming Initialization
def kaiming(in_features, out_features):
    std = np.sqrt(2 / in_features)
    return torch.randn(in_features, out_features) * std
```

```python
# Choose Initialization
weights = [xavier(size, size) for _ in range(layers)]
# weights = [kaiming(size, size) for _ in range(layers)]

activation_mean = []
activation_variance = []

A = X
```

```python
# Forward Pass through 20 Layers
for W in weights:
    A = torch.relu(A @ W)

    activation_mean.append(torch.mean(A).item())
    activation_variance.append(torch.var(A).item())
```

```python
# Print Results
print("Activation Mean")
print(activation_mean)

print("\nActivation Variance")
print(activation_variance)
```

```python
# Plot Mean
plt.plot(activation_mean)
plt.title("Activation Mean")
plt.xlabel("Layer")
plt.ylabel("Mean")
plt.show()
```

```python
# Plot Variance
plt.plot(activation_variance)
plt.title("Activation Variance")
plt.xlabel("Layer")
plt.ylabel("Variance")
plt.show()
```
