```python
#task9
import torch
import torch.nn as nn

# Predicted Bounding Boxes (x, y, w, h)
pred_box = torch.tensor([[50.,60.,100.,120.]])

# Ground Truth Bounding Box
true_box = torch.tensor([[55.,65.,95.,115.]])

# Predicted Class Probability
pred_class = torch.tensor([[0.8]])

# Ground Truth Classa
true_class = torch.tensor([[1.0]])

# Bounding Box Loss
bbox_loss = torch.mean((pred_box - true_box) ** 2)

# Classification Loss
bce = nn.BCELoss()
class_loss = bce(pred_class, true_class)

# Total Loss
total_loss = bbox_loss + class_loss

print("Bounding Box Loss :", bbox_loss.item())
print("Classification Loss :", class_loss.item())
print("Total Loss :", total_loss.item())
```
