```python
#task6
import torch

# Sample Input
torch.manual_seed(42)

X = torch.randn(2,5,3)   # Batch=2, Sequence=5, Features=3

input_size = 3
hidden_size = 4
```

```python
# Initialize Weights
Wi = torch.randn(input_size, hidden_size)
Wf = torch.randn(input_size, hidden_size)
Wo = torch.randn(input_size, hidden_size)
Wc = torch.randn(input_size, hidden_size)
```

```python
# Forward LSTM
h = torch.zeros(2, hidden_size)
c = torch.zeros(2, hidden_size)

forward_output = []

for t in range(5):

    x = X[:,t,:]

    i = torch.sigmoid(x @ Wi)
    f = torch.sigmoid(x @ Wf)
    o = torch.sigmoid(x @ Wo)
    g = torch.tanh(x @ Wc)

    c = f*c + i*g
    h = o*torch.tanh(c)

    forward_output.append(h)

forward_output = torch.stack(forward_output, dim=1)
```

```python
# Backward LSTM
h = torch.zeros(2, hidden_size)
c = torch.zeros(2, hidden_size)

backward_output = []

for t in reversed(range(5)):

    x = X[:,t,:]

    i = torch.sigmoid(x @ Wi)
    f = torch.sigmoid(x @ Wf)
    o = torch.sigmoid(x @ Wo)
    g = torch.tanh(x @ Wc)

    c = f*c + i*g
    h = o*torch.tanh(c)

    backward_output.insert(0,h)

backward_output = torch.stack(backward_output, dim=1)
```

```python
# BiLSTM Output
output = torch.cat((forward_output, backward_output), dim=2)

print("Forward Output Shape :", forward_output.shape)
print("Backward Output Shape:", backward_output.shape)
print("BiLSTM Output Shape  :", output.shape)
```
