```python
#task4
import torch
```

```python
# Sample Data
X = torch.tensor([[1.0],[2.0],[3.0],[4.0]])
Y = torch.tensor([[3.0],[5.0],[7.0],[9.0]])   # y = 2x + 1
```

```python
# Initialize Weight and Bias
W = torch.randn(1)
b = torch.randn(1)

learning_rate = 0.01
epochs = 100
```

```python
# Choose Optimizer
optimizer = "adam"      # gd, momentum, rmsprop, adam
```

```python
# Variables
velocity_w = torch.zeros(1)
velocity_b = torch.zeros(1)

s_w = torch.zeros(1)
s_b = torch.zeros(1)

m_w = torch.zeros(1)
m_b = torch.zeros(1)

v_w = torch.zeros(1)
v_b = torch.zeros(1)

beta1 = 0.9
beta2 = 0.999
eps = 1e-8
```

```python
for epoch in range(1, epochs + 1):

    # Forward Pass
    pred = X * W + b
    loss = torch.mean((pred - Y) ** 2)

    # Gradients
    dW = torch.mean(2 * (pred - Y) * X)
    db = torch.mean(2 * (pred - Y))

    # Gradient Descent
    if optimizer == "gd":
        W = W - learning_rate * dW
        b = b - learning_rate * db

    # Momentum
    elif optimizer == "momentum":
        velocity_w = beta1 * velocity_w + learning_rate * dW
        velocity_b = beta1 * velocity_b + learning_rate * db

        W = W - velocity_w
        b = b - velocity_b

    # RMSprop
    elif optimizer == "rmsprop":
        s_w = beta1 * s_w + (1 - beta1) * dW**2
        s_b = beta1 * s_b + (1 - beta1) * db**2

        W = W - learning_rate * dW / (torch.sqrt(s_w) + eps)
        b = b - learning_rate * db / (torch.sqrt(s_b) + eps)

    # Adam
    elif optimizer == "adam":
        m_w = beta1 * m_w + (1 - beta1) * dW
        m_b = beta1 * m_b + (1 - beta1) * db

        v_w = beta2 * v_w + (1 - beta2) * dW**2
        v_b = beta2 * v_b + (1 - beta2) * db**2

        m_w_hat = m_w / (1 - beta1**epoch)
        m_b_hat = m_b / (1 - beta1**epoch)

        v_w_hat = v_w / (1 - beta2**epoch)
        v_b_hat = v_b / (1 - beta2**epoch)

        W = W - learning_rate * m_w_hat / (torch.sqrt(v_w_hat) + eps)
        b = b - learning_rate * m_b_hat / (torch.sqrt(v_b_hat) + eps)
```

```python
# Result
print("Weight:", W.item())
print("Bias:", b.item())
print("Loss:", loss.item())
```
