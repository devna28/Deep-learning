```python
#task5
import torch

# Sample Input
torch.manual_seed(42)
X = torch.randn(4,3)

gamma = torch.ones(3)
beta = torch.zeros(3)

eps = 1e-5
```

```python
# Batch Normalization
mean = torch.mean(X, dim=0)
variance = torch.var(X, dim=0, unbiased=False)

X_hat = (X - mean) / torch.sqrt(variance + eps)

BN_output = gamma * X_hat + beta

print("Batch Normalization Output")
print(BN_output)
```

```python
# Layer Normalization

mean_ln = torch.mean(X, dim=1, keepdim=True)
variance_ln = torch.var(X, dim=1, keepdim=True, unbiased=False)

X_hat_ln = (X - mean_ln) / torch.sqrt(variance_ln + eps)

LN_output = gamma * X_hat_ln + beta

print("\nLayer Normalization Output")
print(LN_output)
```

```python
# Backward Propagation
dout = torch.ones_like(X)

# Gradients for BatchNorm
dgamma = torch.sum(dout * X_hat, dim=0)
dbeta = torch.sum(dout, dim=0)
```

```python
print("\nBatchNorm Gradients")
print("dGamma =", dgamma)
print("dBeta =", dbeta)
```

```python
# Gradients for LayerNorm
dgamma_ln = torch.sum(dout * X_hat_ln, dim=0)
dbeta_ln = torch.sum(dout, dim=0)
```

```python
print("\nLayerNorm Gradients")
print("dGamma =", dgamma_ln)
print("dBeta =", dbeta_ln)
```
